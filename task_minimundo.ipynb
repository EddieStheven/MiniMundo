# Library

```python
from faker import Faker
import pandas as pd
import random
from sqlalchemy import create_engine
```

```python
fake = Faker('pt_BR')
total_alunos_responsaveis = 2000
```

# Alunos e Responsáveis

* Tabela Alunos 

```python
def gerar_alunos():
    sexo = random.choice(['M', 'F']) 

    if sexo == 'M':
        nome = f"{fake.first_name_male()} {fake.last_name_male()}"  
    else: 
        nome = f"{fake.first_name_female()} {fake.last_name_female()}"  

    return {
        'Nome': nome,
        'Sexo': sexo,
        'Data_Nascimento': fake.date_of_birth(minimum_age=6, maximum_age=18),
        'cep': fake.postcode(),
        'cpf': fake.cpf(),
        'ativo': True 
    }
```

```python
Alunos = [gerar_alunos() for _ in range(total_alunos_responsaveis)]
```

```python
df_Alunos = pd.DataFrame(Alunos)

df_Alunos
```

* Tabela Responsaveis

```python
def gerar_responsaveis():
    sexo = random.choice(['M', 'F']) 

    if sexo == 'M':
        nome = f"{fake.first_name_male()} {fake.last_name_male()}"  
    else: 
        nome = f"{fake.first_name_female()} {fake.last_name_female()}"  

    return {
        'Nome': nome,
        'ativo': True 
    }
```

```python
Responsaveis = [gerar_responsaveis() for _ in range(total_alunos_responsaveis)]
```

```python
df_Responsaveis = pd.DataFrame(Responsaveis)
```

* Criação dos index, tabela Aluno_Responsaveis e organização das tabelas

```python
df_Alunos['id_aluno'] = df_Alunos.index + 1
df_Responsaveis['id_responsavel'] = df_Responsaveis.index + 1
```

```python
df_aluno_responsavel = pd.DataFrame({'id_aluno' : df_Alunos['id_aluno'], 
                                      'id_responsavel' : df_Responsaveis['id_responsavel'].sample(n=len(df_Alunos), replace=True).values})
```

```python
df_Alunos = df_Alunos[['id_aluno', 'Nome', 'Sexo', 'Data_Nascimento', 'cep', 'cpf', 'ativo']]
df_Responsaveis = df_Responsaveis[['id_responsavel','Nome','ativo']]
```
